```python
import carla
```

```python
XODR_FILENAME = "../../scenes/roundabout-saro/maps/SaroRound_2lane_u_adj_14.xodr"
with open(XODR_FILENAME, "r") as f:
    xodr_content = f.read()
map = carla.Map("map", xodr_content)
topology = map.get_topology()

print(f"Number of nodes: {len(topology) * 2}")
```

```python
for entry, exit in topology:
    print(
        f"Entry: {entry.road_id}, {entry.section_id}, {entry.lane_id} -> Exit: {exit.road_id}, {exit.section_id}, {exit.lane_id}"
    )
```
