```python
import omega_prime
from matplotlib import pyplot as plt
```

```python
# Create an Omega Prime recording from an OSI file and an OpenDRIVE map
input_osi_path = "test.osi"
input_xodr_path = "test.xodr"
r = omega_prime.Recording.from_file(input_osi_path, input_xodr_path, validate=True, parse_map=True)
```

## Visualize the $\Omega'$ OmegaPrime file

```python
# Simple plot with specified axis limits

ax = r.plot()
ax.set_xlim(150, 250)
ax.set_ylim(-100, 0)
```

```python
# Video-style plot of the trajectory

r.plot_altair(plot_wedges=False)
```

## Store as $\Omega'$ OmegaPrime

```python
# Store as OpenDrive mcap
r.to_mcap(input_osi_path.replace(".osi", ".mcap"))
```
